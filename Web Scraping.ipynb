```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup 
import smtplib
import time
import datetime
import requests
from urllib.parse import urljoin
```

```python
url = 'https://www.amazon.in/s?k=smart+phones&i=electronics&rh=n%3A1389401031%2Cp_36%3A100000-3530000' # Filtered URL

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Referer": "https://www.google.com/",
    "DNT": "1",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1",
}
```

```python
page = requests.get(url, headers=headers)
soup = BeautifulSoup(page.content, 'html.parser')
links = soup.select("a.a-link-normal.s-no-outline")

page
```

```python
device_title = []
device_price = []
device_rating = []

page_number = 1
while url:
    print(f'Scraping Page {page_number} : {url}')
    page = requests.get(url=url, headers=headers)
    soup = BeautifulSoup(page.content, 'html.parser')

    links = soup.select("a.a-link-normal.s-no-outline")

    for link in links:
        href = link.get('href')

        if not href:
            continue

        product_link = urljoin("https://www.amazon.in", href)
        

        product_response = requests.get(product_link, headers=headers)

        product_soup = BeautifulSoup(product_response.content, 'html.parser')

        title_element = product_soup.find("span", id="productTitle")
        price_element = product_soup.select_one("span.a-price span.a-offscreen")
        rating_element = product_soup.select_one("div#averageCustomerReviews span.a-icon-alt")


        title = title_element.get_text(strip=True) if title_element else None
        price = price_element.get_text(strip=True) if price_element else None
        rating = rating_element.get_text(strip=True) if rating_element else None


        if title:
            device_title.append(title)
            device_price.append(price)
            device_rating.append(rating)

    next_button = soup.select_one("a.s-pagination-next")

    
    if next_button and next_button.get('href'):
        url = urljoin( "https://www.amazon.in", next_button["href"])
        page_number += 1

    else:
        url = None

print("Finished")
print("Products:", len(device_title))
```
