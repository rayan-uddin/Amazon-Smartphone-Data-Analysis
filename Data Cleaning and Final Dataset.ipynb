```python
# Importing modules
import pandas as pd
import numpy as np
```

```python
# Reading the dataset 1
df = pd.read_csv('Data/amazon_scraped_backup.csv')
df
```

```python
# Reading the dataset 2
df2 = pd.read_csv('Data/amazon_scraped_backup_2.csv')
df2
```

```python
# Checking the data types
df.dtypes
```

```python
# Checking the data types
df2.dtypes
```

```python
# Changing the data type of 'Price' Column
df['Price'] = df['Price'].str.replace('₹', '', regex=False).str.replace(',', "", regex=False).astype('float')
```

```python
# Changing the data type of 'Price' Column
df2['Price'] = df2['Price'].str.replace("₹", '',regex=False).str.replace(",", "", regex=False).astype(float)
```

```python
# Changing the data type of 'Rating' Column
df['Rating'] = df['Rating'].str.split().str[0].astype(float)
```

```python
# Changing the data type of 'Rating' Column
df2['Rating'] = df2['Rating'].str.split().str[0].astype(float)
```

```python
# Checking for Devices having Price more than 35k
df[df['Price'] > 35000].sort_values('Price', ascending=False)
```

```python
# Checking for Devices having Price more than 35k
df2[df2['Price'] > 35000].sort_values('Price', ascending=False)
```

```python
# Exporting the cleansed Dataset to .csv
df.to_csv('Data/Cleansed_df.csv', index=False)
```

```python
# Exporting the cleansed Dataset to .csv
df2.to_csv('Data/Cleansed_df2.csv', index=False)
```
